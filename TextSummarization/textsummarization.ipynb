# 🧠 LangChain Summarization Techniques (Theory Only)

LangChain provides multiple built-in summarization strategies for handling documents of various lengths and types. These techniques use large language models (LLMs) to produce summaries by structuring how the text is processed and condensed.

---

## 1. 📝 Stuff Summarization

### Overview:
- In **stuff** summarization, all the content is **"stuffed" into a single prompt** and passed to the LLM to summarize.
- This technique works only when the entire document fits within the token limit of the model.

### Use Case:
- Short articles, blog posts, or emails that do not exceed token limits.

### Advantages:
- Simple and fast.
- Ideal for small-sized inputs.

### Limitations:
- Cannot handle large documents.
- Quality can degrade if the input is too dense.

---

## 2. 🧩 Map-Reduce Summarization

### Overview:
- Inspired by the **MapReduce** programming paradigm.
- First, the document is **split into smaller chunks**.
- Each chunk is summarized independently in the **Map** step.
- These intermediate summaries are then combined and summarized again in the **Reduce** step to form the final output.

### Use Case:
- Long documents or collections of documents.
- When parallelization and scalability are important.

### Advantages:
- Can handle very large inputs.
- Summary is generally more concise than the refine approach.

### Limitations:
- Final summary quality is dependent on intermediate summaries.
- Might lose some contextual continuity between chunks.

---

## 3. 🔁 Refine Summarization

### Overview:
- Starts by summarizing the **first chunk** of the document.
- Then, each subsequent chunk is processed **sequentially**, with the LLM **refining the previous summary** based on new information.

### Use Case:
- Narrative documents or reports where **progressive understanding** is important.
- Tasks requiring high-level continuity and contextual awareness.

### Advantages:
- Maintains coherence across the document.
- Can result in highly contextual summaries.

### Limitations:
- Slower than other methods due to sequential processing.
- Early summary content may overly influence final output.

---

## 🧮 Comparison Table

| Technique   | Best For                      | Handles Long Docs | Coherence | Speed     |
|-------------|-------------------------------|-------------------|-----------|-----------|
| Stuff       | Short documents               | ❌                | High      | Fast      |
| Map-Reduce  | Large docs, scalable systems  | ✅                | Moderate  | Moderate  |
| Refine      | Sequential/narrative content  | ✅                | High      | Slower    |

---

## 🧠 Summary

- Choose **Stuff** for simple, short summaries.
- Use **Map-Reduce** for large-scale summarization with parallelism.
- Opt for **Refine** when the document requires evolving context and coherence.

```python
#prompt template text summarization
from langchain import PromptTemplate
from langchain.chains import LLMChain

generictemplate = """
Write a summary of the follwing speech:
Speech :{speech}
Translate the precise summary to {language}
"""

prompt = PromptTemplate(
    input_variable = ['speech','language'],
    template = generictemplate
)
prompt
```

```python
complete_prompt = prompt.format(speech=speech,language="Hindi")
```

```python
llm_chain = LLMChain(llm=llm,prompt=prompt)
summary = llm_chain.run({'speech':speech,'language':'hindi'})
summary
```

### StuffDocumentChain Text summarization

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("apjspeech.pdf")
docs = loader.load_and_split()
docs
```

```python
template = """Write a concise and short summary of the following speech:
Speech : {text}
"""
prompt = PromptTemplate(input_variables = ['text'],template=template)
```

```python
from langchain.chains.summarize import load_summarize_chain
```

```python
chain = load_summarize_chain(llm,chain_type='stuff',prompt=prompt,verbose=True)
output_summary = chain.run(docs)
output_summary
```

### Map reduce to summarize large documents

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=2000,chunk_overlap=100)
final_document = text_splitter.split_documents(docs)
final_document
```

```python
chunk_prompt = """
Please summarize the below speech:
Speech : {text}
Summary:
"""

map_prompt_template = PromptTemplate(input_variables=['text'], template=chunk_prompt)
```

```python
final_prompt = '''
Provide the final summary of the entire speech with these important points:
Add a motivational title, Start the precise summary with an introduction and provide the summary in number of points
for the speech.
Speech: {text}
'''
final_prompt_template = PromptTemplate(input_variables=['text'],template=final_prompt)
```

```python
summary_chain = load_summarize_chain(
    llm=llm,
    chain_type='map_reduce',
    map_prompt=map_prompt_template,
    combine_prompt=final_prompt_template,
    verbose=True
)
```

```python
output = summary_chain.run(final_document)
```

```python
#refine 
chain = load_summarize_chain(
    llm=llm,
    chain_type="refine",
    verbose=True
)
outputSummary = chain.run(final_document)
outputSummary
```
