```python
from langchain.vectorstores.cassandra import Cassandra
from langchain_ollama import OllamaEmbeddings
from langchain_groq import ChatGroq
from langchain.indexes.vectorstore import VectorStoreIndexWrapper

#support for dataset retrieval using HuggingFace
from datasets import load_dataset

#cassio is used as an engine powering cassandra/astradb integration in langchain
#also used for DB connection
import cassio

#pdfreader
from PyPDF2 import PdfReader
```

```python
#Setup
import os
from dotenv import load_dotenv
load_dotenv()

ASTRA_DB_APPLICATION_TOKEN = os.getenv("ASTRA_DB_APPLICATION_TOKEN")
ASTRA_DB_ID = os.getenv("ASTRA_DB_ID")
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
pdfreader = PdfReader("budget_speech.pdf")
```

```python
from typing_extensions import Concatenate

raw_text = ''
for i, page in enumerate(pdfreader.pages):
    content = page.extract_text()
    if content:
        raw_text += content
```

```python
raw_text
```

```python
#initializing connection to database
cassio.init(token=ASTRA_DB_APPLICATION_TOKEN,database_id=ASTRA_DB_ID)
```

```python
#create langchain llm and embeddings objects
llm = ChatGroq(groq_api_key = GROQ_API_KEY, model = "openai/gpt-oss-120b")
embeddings = OllamaEmbeddings(model="llama3.1:latest")
```

```python
#creating langchain vector store using astra db
vector_store = Cassandra(
    embedding=embeddings,
    table_name="pdf_table",
    session=None,
    keyspace=None
)
```

```python
from langchain.text_splitter import CharacterTextSplitter
text_splitter = CharacterTextSplitter(
    separator = "\n",
    chunk_size = 800,
    chunk_overlap = 150,
    length_function = len
)
texts = text_splitter.split_text(raw_text)
texts
```

```python
texts[:50]
```

```python
vector_store.add_texts(texts[:50])
print("Inserted %i headlines." % len(texts[:50]))
vector_index = VectorStoreIndexWrapper(vectorstore=vector_store)
```

```python
first_question = True
while True:
    if first_question:
        query_text = input("\nEnter your question (or type 'quit' to exit): ").strip()
    else:
        query_text = input("\nWhat's your next question (or type 'quit' to exit): ").strip()

    if query_text.lower() == "quit":
        break

    if query_text == "":
        continue

    first_question = False

    print("\nQUESTION: \"%s\"" % query_text)
    answer = vector_index.query(query_text, llm=llm).strip()
    print("ANSWER: \"%s\"\n" % answer)

    print("FIRST DOCUMENTS BY RELEVANCE:")
    for doc, score in vector_store.similarity_search_with_score(query_text, k=4):
        print("    [%0.4f] \"%s ...\"" % (score, doc.page_content[:84]))
```
