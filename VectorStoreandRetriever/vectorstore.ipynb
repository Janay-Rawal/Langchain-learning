# 📄 `langchain_core.documents` — LangChain Document Class Explained

## 📦 What is It?

`langchain_core.documents` provides the `Document` class — a lightweight wrapper to store a piece of text along with its metadata.

This is commonly used in:
- Text chunking and splitting
- Vectorstore storage (like FAISS, Pinecone, Chroma)
- Retrieval-Augmented Generation (RAG)
- Passing source documents into chains

````python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq

groq_api_key = os.getenv("GROQ_API_KEY")

os.environ["HF_API_TOKEN"] = os.getenv("HF_API_TOKEN")

llm = ChatGroq(groq_api_key=groq_api_key,model="llama-3.1-8b-instant")
llm
````

````python
from langchain_core.documents import Document


documents = [
    Document(
        page_content="Dogs are loyal mammals often considered man's best friend. They require daily exercise and companionship.",
        metadata={"type": "mammals-pets-doc", "animal": "dog"}
    ),
    Document(
        page_content="Cats are independent mammals known for their agility and cleanliness. They often enjoy sleeping in sunny spots.",
        metadata={"type": "mammals-pets-doc", "animal": "cat"}
    ),
    Document(
        page_content="Goldfish are small freshwater fish that are commonly kept in bowls or aquariums. They are peaceful and low-maintenance.",
        metadata={"type": "fish-pets-doc", "animal": "goldfish"}
    ),
    Document(
        page_content="Parrots are intelligent birds known for their ability to mimic human speech. They require social interaction and mental stimulation.",
        metadata={"type": "birds-pets-doc", "animal": "parrot"}
    ),
    Document(
        page_content="Golden Retrievers are large, friendly dogs with a love for water and family activities. They are excellent with children.",
        metadata={"type": "mammals-pets-doc", "animal": "dog"}
    ),
    Document(
        page_content="Siamese cats are a vocal and affectionate breed, easily recognizable by their distinct markings and blue eyes.",
        metadata={"type": "mammals-pets-doc", "animal": "cat"}
    ),
    Document(
        page_content="Comet goldfish are fast swimmers and require more space than typical bowl setups. They thrive in filtered tanks.",
        metadata={"type": "fish-pets-doc", "animal": "goldfish"}
    ),
    Document(
        page_content="Macaws are a colorful species of parrot with long tails and strong beaks. They bond closely with their caretakers.",
        metadata={"type": "birds-pets-doc", "animal": "parrot"}
    )
]

````

````python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model="all-MiniLM-L6-v2")
````

````python
#VectorStore

from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(documents,embedding=embeddings)
vectorstore
````

````python
vectorstore.similarity_search("cat")
````

````python
#Async query
await vectorstore.asimilarity_search("cat")
````

````python
vectorstore.similarity_search_with_score("cat")
````

````python
from typing import List
from langchain_core.runnables import RunnableLambda

retriever = RunnableLambda(vectorstore.similarity_search).bind(k=1)
retriever.batch(["cat"])
````

````python
##using as_retriever (better practice)

retriever = vectorstore.as_retriever(
    search_type = "similarity",
    search_kwargs = {"k":1}
)
retriever.batch(["cat"])
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough

message = """
Answer this question using the provided context only.
{question}

Context:
{context}
"""

prompt = ChatPromptTemplate.from_messages([("human",message)])
rag_chain = {"context":retriever,"question":RunnablePassthrough()}|prompt|llm

response = rag_chain.invoke("Tell me about dogs")

response
````
