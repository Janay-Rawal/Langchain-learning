```python
from langchain_ollama import OllamaEmbeddings
```

```python
embeddings = (
    OllamaEmbeddings(model = "llama3.2:1b")  #by default it uses llama2
)
```

```python
embeddings
```

```python
r1 = embeddings.embed_documents(
    [
    'Alpha is the first character of the Greek alphabet.',
    'Beta is the second letter of the Greek alphabet.'
    ]
)
```

```python
len(r1[0])
```

```python
r1[0]
```

```python
embeddings.embed_query("What is the first letter of the Greek alphabet?")
```

```python
doc_result = embeddings.embed_documents([r1, "this is not a test document"])
```
