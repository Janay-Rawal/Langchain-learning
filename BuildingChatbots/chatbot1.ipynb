# 📚 Building a Chatbot With Message History Using LangChain

## 🧠 Introduction

This lecture focuses on building a chatbot that can remember previous interactions using **message history**.  
The chatbot uses **LangChain** and connects to a **Groq-hosted language model** instead of OpenAI.  
The key idea is to make the bot capable of maintaining context across conversations, especially for **different users** at the same time.

---

## ⚙️ Environment Setup

The environment is prepared by:

- Importing necessary Python libraries  
- Loading environment variables (like API keys)  
- Retrieving the **Groq API key** to access the language model  

---

## 🧠 Loading the Language Model

An **LLM (Language Model)** is loaded from Groq using LangChain's integration.  
The model used is `gama-2-9b`, which is a fast and powerful open-source chat model.  
This model will act as the chatbot engine that understands and replies to user input.

---

## 💬 Basic Conversation

Initially, the chatbot handles a **one-turn conversation**.  
A user message is sent, and the model replies.  
However, the model **does not retain any memory** of what was said in previous messages.  
It only knows what is currently being asked.

---

## 🔁 Maintaining Context Temporarily

To simulate a multi-turn conversation, a **list of messages** is manually provided to the model.  
This allows it to respond with some context, such as remembering the user's name.  
But this memory **only lasts for that single function call** — it is not retained automatically across multiple interactions.

---

## 🧱 Introduction to Message History

In real-world applications, chatbots need to:

- Remember what the user said earlier  
- Handle **multiple users or sessions** at the same time  
- Provide different memory contexts per user  

This is where **message history** and **session management** become essential.  
Each session should have its **own chat memory** so that context doesn't get mixed between users.

---

## 🧰 Message History Libraries

To implement message history, LangChain provides several tools:

- **ChatMessageHistory**: A built-in way to store past messages temporarily (in memory).
- **BaseChatMessageHistory**: A base class used when creating custom storage solutions.
- **RunnableWithMessageHistory**: A wrapper that adds memory support to your chatbot or agent.

These tools help LangChain remember what was said and by whom.

---

## 🆔 Using Session IDs

Each chat session is given a unique **session ID** like `"chat1"` or `"user42"`.  
This ID links the user to their specific chat history.  
When a message is sent with a session ID, the chatbot will retrieve and use that session’s history — maintaining continuity in the conversation.

---

## 🔄 Multi-Session Demonstration

When using different session IDs:

- A new session will **not remember** anything from another session.
- If the same session ID is used again, the chatbot will **retain and use previous context**.

For example:

- If in one session you say “My name is Krish”, and later ask “What’s my name?”, the bot will remember.
- But if you switch to a new session ID and ask the same question, the bot won’t know — until you tell it again.

---

## ✅ Summary of Key Concepts

- **LangChain** allows building intelligent chatbots using modular components.
- **Message history** lets your chatbot retain memory between messages.
- **Session IDs** make it possible to support multiple users simultaneously, each with their own private conversation memory.
- **ChatMessageHistory** stores messages in memory (can be replaced with Redis, DB, etc.).
- **RunnableWithMessageHistory** connects your chatbot model with the memory system.
- This design makes it easy to manage and switch between conversations in real-world applications.

# Message history: ChatMessageHistory, BaseChatMessageHistory, RunnableWithMessageHistory

Brief overview:

- **ChatMessageHistory**: a concrete, in-memory implementation for storing chat messages (user + AI) for a session. Use it when you need a simple message store during development or short-lived sessions.
- **BaseChatMessageHistory**: an abstract base class (interface) describing the methods a chat message history must implement. Use this when you want to implement a custom storage backend (database, Redis, etc.) that integrates with LangChain's history expectations.
- **RunnableWithMessageHistory**: a small wrapper that composes a runnable (for example, a model or chain) and automatically records inputs and outputs into a message history. Use it when you want message logging for each invocation without adding persistence logic to your application code.

When to pick which:

- For quick local tests or demos: **ChatMessageHistory**.
- For production or cross-process state: implement **BaseChatMessageHistory** and back it with a DB or cache.
- To transparently log every model/chain call: wrap the runnable with **RunnableWithMessageHistory**.

The code cell below shows a minimal example: create/get a session history, add messages, inspect messages, and wrap a model runnable so invocations are recorded automatically.

```python
#Message history
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}

def get_session_history(session_id:str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model,get_session_history)
```

```python
config = {"configurable" : {"session_id" : "chat1"}}
```

```python
response = with_message_history.invoke(
    [
        HumanMessage(content = "Hello, I am Janay Rawal and I am learning Langchain.")
    ],
    config = config
)
response.content
```

```python
config2 = {"configurable" : {"session_id" : "chat2"}}

response = with_message_history.invoke(
    [
        HumanMessage(content = "Hello, I am Sanvi Verma and I am learning Langchain.")
    ],
    config = config2
)
response.content
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are a helpful assistant. Answer all the questions to the best of your ability"),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain = prompt|model
```

```python
response = chain.invoke(
    {
        "messages":[HumanMessage(content="Hi! I am Janay")]
    }
)
response
```

```python
with_message_history = RunnableWithMessageHistory(chain,get_session_history)
config = {"configurable": {"session_id" : "chat3"}}

response = with_message_history.invoke(
    [HumanMessage(content="Hi! I am Janay")],
    config=config
)
response
```

```python
##more complexity

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are a helpful assistant. Answer all the questions to the best of your ability in {language}."),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain = prompt|model



response = chain.invoke(
    {
        "messages" : [HumanMessage(content="Hi! My name is Janay")],
        "language" : "Hindi"

    }
)

response
```

```python
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
    )
config = {"configurable" : {"session_id" : "chat3"}}

response = with_message_history.invoke(
    {"messages" : [HumanMessage(content = "Hi! my age is 20")],
     "language" : "Hindi"},
     config=config
)
response
```
